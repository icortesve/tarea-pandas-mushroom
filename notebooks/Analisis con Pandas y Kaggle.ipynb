### Desarrollo

Lo primero es usar una celda para traducir los "códigos"  utilzados en el dataframe. Sólo se realiza con las columnas objetivo.

```python
#  Diccionarios de traducción para las columnas objetivo
traduccion_clase = {
    'p': 'Venenoso', 
    'e': 'Comestible'
}

traduccion_olor = {
    'a': 'Almendra',
    'l': 'Anís',
    'c': 'Parafina',
    'y': 'Pescado',
    'f': 'Fétido',
    'm': 'Moho',
    'n': 'Sin olor',
    'p': 'Tipo cloro',
    's': 'Aliños'
}

traduccion_esporas = {
    'k': 'Negro',
    'n': 'Marrón',
    'b': 'Beige',
    'h': 'Chocolate',
    'r': 'Verde',
    'o': 'Naranja',
    'u': 'Morado',
    'w': 'Blanco',
    'y': 'Amarillo'
}
```

```python
# Reemplazar las traducciones en cada columna
df_limpio['class'] = df_limpio['class'].replace(traduccion_clase)
df_limpio['odor'] = df_limpio['odor'].replace(traduccion_olor)
df_limpio['spore-print-color'] = df_limpio['spore-print-color'].replace(traduccion_esporas)

print("¡Traducción completada con éxito!")

# Primeras filas para verificar el resultado
df_limpio[['class', 'odor', 'spore-print-color']].head()
```

#### --- Análisis Base del Data frame --

```python
# Contar hongos por clase
conteo_clases = df_limpio['class'].value_counts()

# Calcular la proporción (porcentaje) de cada clase respecto al total
proporcion_clases = df_limpio['class'].value_counts(normalize=True) * 100

# Combinar ambos resultados en un DataFrame limpioal
reporte_clases = pd.DataFrame({
    'Cantidad Absoluta': conteo_clases,
    'Porcentaje (%)': proporcion_clases
})

# Se definen los valores de la primera columna
reporte_clases.index = reporte_clases.index.map({
    'Comestible': 'Comestible (e)', 
    'Venenoso': 'Venenoso (p)'
})

print("--- BALANCE GENERAL DEL DATASET ---")
print(reporte_clases.round(2))
```

 ### HIPÓTESIS 1: El olor es el mecanismo biológico de advertencia más fuerte.

1.1 Objetivo: Contar cuántos hongos comestibles y venenosos hay por cada olor.

```python
# Obtener los valores únicos de la columna 'odor'
olores_unicos = df_limpio['odor'].unique()

print(f"Se encontraron {len(olores_unicos)} tipos diferentes de olor en el dataset.")
print("Los aromas identificados son:", olores_unicos)
```

```python
# Contar hongos por cada olor
conteo_por_olor = df_limpio['odor'].value_counts()

print("--- CANTIDAD DE HONGOS SEGÚN SU AROMA ---")
print(conteo_por_olor)
```

Cruce de Olor vs. Clase (Comestible / Venenoso)
 
1.2 Objetivo: Ver si existen aromas que sean exclusivos de hongos venenosos o comestibles.

```python
# Agrupar por olor y clase, y contarcuántos hongos hay en cada combinación
perfil_olor = df_limpio.groupby(['odor', 'class']).size().unstack(fill_value=0)

# Renombrar columnas
perfil_olor.columns = ['Comestible', 'Venenoso']

print("--- TABLA AROMA VS. CLASE ---")
print(perfil_olor)
```

Agrupar los hongos en 'Grupos Sensoriales' en base a si su olor característico es riesgoso o no

```python
# 1. Calculamos los porcentajes exactos por grupo
p_quimicos = df_limpio['odor'].isin(['Fétido', 'Tipo cloro', 'Parafina', 'Pescado', 'Aliños', 'Moho']).mean() * 100
p_dulces = df_limpio['odor'].isin(['Almendra', 'Anís']).mean() * 100
p_sin_olor = (df_limpio['odor'] == 'Sin olor').mean() * 100

# 2. Los metemos directamente en una tabla estructurada
tabla_resumen = pd.DataFrame({
    'Porcentaje (%)': [p_quimicos, p_dulces, p_sin_olor]
}, index=['Químico/Fétido', 'Dulce (Anís/Almendra)', 'Sin Olor'])

print("--- RESUMEN DE GRUPOS SENSORIALES ---")
print(tabla_resumen.round(2).astype(str) + '%')
```

Cálculo de Porcentaje de riesgo por olor

1.3 Objetivo: Traducir los conteos anteriores en una métrica matemática de riesgo.


```python
# Recorrer fila por fila (axis=1) y calcular el porcentaje
perfil_olor['% de Riesgo Mortal'] = perfil_olor.apply(
    lambda fila: (fila['Venenoso'] / (fila['Comestible'] + fila['Venenoso'])) * 100,
    axis=1
)

print("--- REPORTE FINAL DE LA HIPÓTESIS 1 (CON % DE RIESGO) ---")
print(perfil_olor.round(2))
```

```python
# ==============================================================================
# SÍNTESIS FINAL: ANÁLISIS DE LA HIPÓTESIS 1 (EL OLOR)
# ==============================================================================

print("\n" + "="*85)
print("SÍNTESIS DE RESULTADOS: HIPÓTESIS 1 (EL OLOR)")
print("="*85)

# 1. DESGLOSE BASADO EN LA TABLA DE RESUMEN
print(
    f"1. RESULTADOS OBTENIDOS:\n"
    f"   • Ítem 1.1 (Distribución): El bosque se divide en tres perfiles sensoriales:\n"
    f"     - Grupo Químico/Fétido: {p_quimicos:.2f}% (Identificado como tóxico).\n"
    f"     - Grupo Dulce: {p_dulces:.2f}% (Identificado como seguro).\n"
    f"     - Grupo Sin Olor: {p_sin_olor:.2f}% (Zona de incertidumbre).\n\n"
    "   • Ítem 1.2 (Exclusividad): Los aromas actúan como marcadores binarios.\n"
    "     No existe ningún olor desagradable en hongos comestibles, ni olores\n"
    "     dulces en hongos venenosos. La separación es del 100%.\n\n"
    "   • Ítem 1.3 (Riesgo): El olfato elimina el riesgo en el 56.6% del dataset.\n"
    "     Sin embargo, deja un riesgo latente del 3.4% dentro del grupo 'Sin olor',\n"
    "     donde se esconden 120 ejemplares venenosos."
)

print("-" * 85)

# 2. CONCLUSIÓN Y DERIVACIÓN A LA HIPÓTESIS 2
print(
    "2. CONCLUSIÓN Y NECESIDAD DE LA H2:\n"
    "   La Hipótesis 1 se VALIDA como un filtro de descarte inicial sumamente\n"
    "   efectivo, pero INSUFICIENTE para garantizar la seguridad total.\n\n"
    "   ¿Por qué pasar a la Hipótesis 2 (Esporas)?:\n"
    "   El grupo 'Sin Olor' es demasiado grande y peligroso para ser ignorado.\n"
    "   Como el olfato no entrega señales en este escenario, es obligatorio\n"
    "   introducir la Hipótesis 2 para analizar si el color de las esporas\n"
    "   puede discriminar a esos 120 hongos venenosos 'mudos' y así cumplir\n"
    "   con el objetivo de Riesgo Cero."
)
print("="*85 + "\n")
```
