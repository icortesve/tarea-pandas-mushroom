### OBJETIVO GENERAL:

### Identificar patrones físicos y firmas biológicas consistentes que permitan diferenciar con certeza un hongo comestible de uno venenoso, reduciendo el riesgo de intoxicación.

```python
# --- Análisis Base del Data frame --

# Contar hongos por clase
conteo_clases = df_limpio['class'].value_counts()

# Calcular la proporción (porcentaje) de cada clase respecto al total
proporcion_clases = df_limpio['class'].value_counts(normalize=True) * 100

# Combinar ambos resultados en un DataFrame limpio para mostrar un reporte formal
reporte_clases = pd.DataFrame({
    'Cantidad Absoluta': conteo_clases,
    'Proporción (%)': proporcion_clases
})

# Cambiar los índices para que el reporte sea legible en español
reporte_clases.index = reporte_clases.index.map({'e': 'Comestible (e)', 'p': 'Venenoso (p)'})

print("--- BALANCE GENERAL DEL DATASET ---")
print(reporte_clases.round(2))
```


6. Análisis de Datos

*   Realiza agrupaciones de datos utilizando groupby para obtener insights específicos (por ejemplo, ventas por producto, ventas por región, etc.).
*   Aplica funciones de agregación como sum, mean, count, min, max, std, y var.
*   Utiliza el método apply para realizar operaciones más complejas y personalizadas.


```python
# ==============================================================================
# DICCIONARIO MAESTRO DE MAPEO (GLOSARIO COMPLETO)
# ==============================================================================

traductor_maestro = {
    'class': {
        'e': 'Comestible', 
        'p': 'Venenoso'
    },
    'odor': {
        'a': 'Almendra', 'l': 'Anís', 'c': 'Creosota', 'y': 'Pescado', 
        'f': 'Fétido', 'm': 'Moho', 'n': 'Sin olor', 'p': 'Pungente', 's': 'Especiado'
    },
    'spore-print-color': {
        'k': 'Negro', 'n': 'Marrón', 'b': 'Gamuza', 'h': 'Chocolate', 
        'r': 'Verde', 'o': 'Naranja', 'u': 'Púrpura', 'w': 'Blanco', 'y': 'Amarillo'
    },
    'cap-color': {
        'n': 'Marrón', 'b': 'Gamuza', 'c': 'Canela', 'g': 'Gris', 'r': 'Verde', 
        'p': 'Rosado', 'u': 'Púrpura', 'e': 'Rojo', 'w': 'Blanco', 'y': 'Amarillo'
    },
    'habitat': {
        'g': 'Pastizales', 'l': 'Hojas', 'm': 'Prados', 'p': 'Senderos', 
        'u': 'Urbano', 'w': 'Desechos', 'd': 'Bosques'
    }
    # Nota: Puedes seguir agregando aquí las demás columnas si tu informe lo requiere
}

# ==============================================================================
# TRADUCCIÓN AUTOMÁTICA DE TODO EL DATASET
# ==============================================================================

# Recorremos cada columna que definimos en nuestro diccionario maestro
for columna, mapeo in traductor_maestro.items():
    if columna in df_limpio.columns:
        df_limpio[columna] = df_limpio[columna].map(mapeo)

print("¡Todo el dataset ha sido traducido al español usando el Diccionario Maestro!")
```

```python
# ------------------------------------------------------------------------------
# HIPÓTESIS 1: El olor es el mecanismo biológico de advertencia más fuerte.
# OBJETIVO: Contar cuántos hongos comestibles y venenosos hay por cada olor.
# ------------------------------------------------------------------------------

# Obtener los valores únicos de la columna 'odor'
olores_unicos = df_limpio['odor'].unique()

print(f"Se encontraron {len(olores_unicos)} tipos diferentes de olores en el dataset.")
print("Los aromas identificados son:", olores_unicos)
```

```python
conteo_por_olor = df_limpio['odor'].value_counts()

print("--- CANTIDAD DE HONGOS SEGÚN SU AROMA ---")
print(conteo_por_olor)
```

```python
# ==============================================================================
# PASO 1.3: Cruce de Olor vs. Clase (Comestible / Venenoso)
# OBJETIVO: Ver si existen aromas que sean exclusivos de hongos venenosos o comestibles.
# ==============================================================================

# Agrupamos por olor y clase, y contamos cuántos hongos hay en cada combinación
perfil_olor = df_limpio.groupby(['odor', 'class']).size().unstack(fill_value=0)

# Renombramos las columnas para que el reporte sea perfectamente legible
perfil_olor.columns = ['Comestible', 'Venenoso']

print("--- TABLA DE CRUCE: AROMA VS. CLASE ---")
print(perfil_olor)
```

```python
# ==============================================================================
# PASO 1.4: Cálculo de Porcentaje de Riesgo por Olor (.apply)
# OBJETIVO: Traducir los conteos anteriores en una métrica matemática de riesgo.
# ==============================================================================

# Usamos .apply() para recorrer fila por fila (axis=1) y calcular el porcentaje
perfil_olor['% de Riesgo Mortal'] = perfil_olor.apply(
    lambda fila: (fila['Venenoso'] / (fila['Comestible'] + fila['Venenoso'])) * 100,
    axis=1
)

print("--- REPORTE FINAL DE LA HIPÓTESIS 1 (CON % DE RIESGO) ---")
print(perfil_olor.round(2))
```

```python
# ==============================================================================
# PASO 1.5: SÍNTESIS CIENTÍFICA DEL ANÁLISIS POR OLOR
# ==============================================================================

print("\n" + "="*80)
print("SÍNTESIS DEL ANÁLISIS DE LA HIPÓTESIS 1 (EL OLOR) RELATIVO AL OBJETIVO GENERAL")
print("="*80)
print(
    "1. VALIDACIÓN DE LA HIPÓTESIS:\n"
    "   La hipótesis se VALIDA COMPLETAMENTE para hongos con aromas perceptibles.\n"
    "   Los olores químicos o desagradables (Fétido, Pungente, Creosota, Pescado,\n"
    "   Especiado y Moho) actúan como una firma física infalible de toxicidad,\n"
    "   registrando un 100.0% de Riesgo Mortal en los datos. Por el contrario,\n"
    "   los aromas dulces (Almendra y Anís) garantizan un 0.0% de riesgo.\n"
)
print(
    "2. APORTE AL OBJETIVO GENERAL (LIMITACIÓN BIOLÓGICA CON 'SIN OLOR'):\n"
    "   Aunque el olfato permite descartar o aceptar el 56.6% de los hongos del\n"
    "   dataset con total certeza, el grupo 'Sin olor' representa una trampa\n"
    "   de recolección. Al tener un 3.4% de riesgo latente (120 hongos venenosos\n"
    "   ocultos), demuestra que el olor NO es suficiente como único criterio.\n"
    "   Para cumplir el objetivo general de seguridad absoluta, es obligatorio\n"
    "   cruzar estos resultados con una segunda hipótesis física (las esporas)."
)
print("="*80 + "\n")
```
